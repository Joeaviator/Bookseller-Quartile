```python
df = pd.read_csv('Bestsellers with categories.csv')
df.head()
```

```python
#Quantilesof age
Price_q1=np.quantile(df['Price'],0.25)
print("First quantile: ", Price_q1)
Price_q2=np.quantile(df['Price'],0.50)
print("First quantile: " ,Price_q2)
Price_q3=np.quantile(df['Price'],0.75)
print("First quantile: ", Price_q3)
```

```python
Inter_QR_P = Price_q3 - Price_q1
print("Inter Quartile Range: ", Inter_QR_P)
```

```python
#Quantilesof age
User_q1=np.quantile(df['User Rating'],0.25)
print("First quantile: ", User_q1)
User_q2=np.quantile(df['User Rating'],0.50)
print("First quantile: " ,User_q2)
User_q3=np.quantile(df['User Rating'],0.75)
print("First quantile: ", User_q3)
```

```python
Inter_QR_U= User_q3 - User_q1
print("Inter Quartile Range: ", Inter_QR_U)
```

```python
sns.boxplot(df['Price'], palette='rainbow', legend=True)
plt.title("Price Distribution")
```

```python
sns.boxplot(df['User Rating'], palette='rainbow', legend=True)
plt.title(" User Rating Distribution")
```
